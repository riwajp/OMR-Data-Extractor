Import the dataset and libraries.

```python
import pandas as pd
import ast
import numpy as np

# Load the CSV file
csv_path = "./labeled-data-1/labeled-data.csv"
df = pd.read_csv(csv_path)

```

Check if any image has less than 6 marks.

```python
print(df[df["count"]<6])
```

Remove marks detection with x>1050 (no circles there, so remove misread marks).

```python
def filter_coords(coord_str):
    coords = np.array(ast.literal_eval(coord_str))
  
    filtered = [tuple(pt) for pt in coords if not (pt[1]<300 and pt[0]>=1100)]
  
   
    return filtered

# Apply filtering and update the dataframe
df['marked'] = df['marked'].apply(filter_coords)

df["count"]=df['marked'].apply(len)

pd.set_option('display.max_colwidth', None)

print(df[df["count"]!=6][["image","marked"]])
bad_images=df[df["count"]!=6 ]["image"]








```

Save images with < 6 marks for review

```python
import shutil
import os

def copy_files(file_names, src_dir, dest_dir):
    print(file_names)
    os.makedirs(dest_dir, exist_ok=True)  # Ensure destination directory exists
    
    for file_name in file_names:
        src_path = os.path.join(src_dir, file_name)
        dest_path = os.path.join(dest_dir, file_name)

        if os.path.exists(src_path):
            shutil.copy2(src_path, dest_path)
            print(f"Copied: {file_name}")
        else:
            print(f"File not found: {file_name}")
source_directory = "./labeled-data-1"
destination_directory = "./bad-data"

copy_files(bad_images, source_directory, destination_directory)
```

Extract set and answers data from the labeled dataset.

```python
from transform import getOption
import csv


data=[]
for idx, row in df.iterrows():
    row_data=getOption(row["image"],(row["green_margin_x"],row["green_margin_y"]),row["marked"],row["left_margin_x"])
    row_data["image"]=row["image"]
    data.append(row_data)

filename = 'answers-data.csv'

with open(filename, mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['image','set', 'data'])  # Write header

    for item in data:
        writer.writerow([item['image'],item['set'], str(item['data'])])

```
