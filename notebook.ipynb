Import the dataset and libraries.

```python
import pandas as pd
import ast
import numpy as np

# Load the CSV file
csv_path = "./marked_binary/mark_positions.csv"
df = pd.read_csv(csv_path)

```

Filter coordinates with x > 900. (No circles beyond that point, so remove the misread marks.)

```python

def filter_coords(coord_str):
    coords = np.array(ast.literal_eval(coord_str))
    filtered = [tuple(pt) for pt in coords if pt[0] <= 900]
    return np.array(filtered)

# Apply filtering and update the dataframe
df['marked'] = df['marked'].apply(filter_coords)
df['count'] = df['marked'].apply(len)

# Optional: Save to a new CSV
filtered_csv_path = "./mark_positions_filtered.csv"
df.to_csv(filtered_csv_path, index=False)

# Show result
print(df.head())
```

Check if any image has more than 6 marks.

```python
print(df[df["count"]>6])
```

Expect circles for the first column between x=[155,438]

```python
def filter_coords(coords):
    
    filtered = [tuple(pt) for pt in coords[1:] if pt[0]>=155 and pt[0] <= 438]
    return np.array(filtered)

# Apply filtering and update the dataframe
df['first_col_marked'] = df['marked'].apply(filter_coords)

df["first_col_count"]=df['first_col_marked'].apply(len)

print(df[["image","first_col_count"]])
```
